```python
# MOVIE RECOMMENDATION SYSTEM
# SPARKKIT INTERNSHIP PROJECT

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Movie dataset
movies = pd.DataFrame({
    "Movie": [
        "Inception",
        "Interstellar",
        "The Dark Knight",
        "Avengers: Endgame",
        "Iron Man",
        "The Avengers",
        "Titanic",
        "Avatar",
        "Jurassic Park",
        "The Matrix",
        "Harry Potter",
        "Toy Story",
        "Finding Nemo",
        "The Lion King",
        "Frozen"
    ],
    
    "Genre": [
        "Science Fiction Thriller",
        "Science Fiction Drama",
        "Action Crime Drama",
        "Action Science Fiction Adventure",
        "Action Science Fiction",
        "Action Science Fiction Adventure",
        "Romance Drama",
        "Science Fiction Adventure",
        "Science Fiction Adventure",
        "Science Fiction Action",
        "Fantasy Adventure",
        "Animation Comedy Family",
        "Animation Adventure Family",
        "Animation Drama Family",
        "Animation Fantasy Family"
    ],
    
    "Description": [
        "A thief enters dreams to steal secrets.",
        "Astronauts travel through space to find a new home.",
        "A superhero fights dangerous criminals.",
        "Superheroes unite to save the world.",
        "A billionaire builds an armored superhero suit.",
        "Superheroes join together to protect Earth.",
        "A romantic story between two people on a ship.",
        "Humans explore an alien world.",
        "Scientists bring dinosaurs back to life.",
        "A hacker discovers a reality controlled by machines.",
        "A young wizard studies magic and fights evil.",
        "Toys come to life and experience adventures.",
        "A fish searches for his missing son.",
        "A young lion learns to become king.",
        "A princess with magical powers saves her kingdom."
    ]
})

# 2. Combine genre and description
movies["Features"] = movies["Genre"] + " " + movies["Description"]

# 3. Convert text into numerical features
vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = vectorizer.fit_transform(movies["Features"])

# 4. Calculate similarity
similarity = cosine_similarity(tfidf_matrix)

# 5. Recommendation function
def recommend(movie_name, number=5):
    
    if movie_name not in movies["Movie"].values:
        print("Movie not found!")
        return
    
    movie_index = movies[movies["Movie"] == movie_name].index[0]
    
    scores = list(enumerate(similarity[movie_index]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)
    
    print("\nRecommendations for:", movie_name)
    print("-" * 45)
    
    for index, score in scores[1:number+1]:
        print(
            movies.iloc[index]["Movie"],
            " | Similarity:",
            round(score, 3)
        )

# 6. Display dataset
print("=" * 60)
print("MOVIE RECOMMENDATION SYSTEM")
print("=" * 60)

print("\nMovie Dataset:")
display(movies[["Movie", "Genre", "Description"]])

# 7. Test recommendation
recommend("Inception", 5)

print("\n")
recommend("Avengers: Endgame", 5)

print("\n")
recommend("Titanic", 5)

# 8. Final message
print("\n" + "=" * 60)
print("PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 60)
print("Method Used: TF-IDF + Cosine Similarity")
print("The system recommends movies based on similar content.")
```
